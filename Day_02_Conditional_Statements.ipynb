## 1. Conditional Statements

Conditional statements are used to make decisions in a Python program based on whether a condition is `True` or `False`.

### Types
1. `if`
2. `if-else`
3. `if-elif-else` (if-else ladder)
4. Nested `if-else`


## 2. Simple `if` Statement

**Syntax:**
```python
if condition:
    statements
```

The statements inside the `if` block execute only when the condition is True.

````python
# Check voting eligibility based on age
age = int(input("Enter age: "))

if age >= 18:
    print("Eligible for voting")
````

## 3. `if-else` Statement

**Syntax:**
```python
if condition:
    statements
else:
    statements
```

The `if` block executes when the condition is True. Otherwise, the `else` block executes.

````python
age = int(input("Enter age: "))

if age >= 18:
    print("Eligible for voting")
else:
    print("Not eligible for voting")
````

## 4. Voting Eligibility Based on Age and Country

The `and` operator requires both conditions to be True.

````python
country = input("Enter country: ")
age = int(input("Enter age: "))

if age >= 18 and country.lower() == "india":
    print("Eligible for voting")
else:
    print("Not eligible for voting")
````

**Note:** `lower()` converts the country name to lowercase. Therefore, inputs such as `India`, `INDIA` and `InDiA` are treated consistently.

## 5. Even or Odd Number

A number is even when its remainder after division by 2 is zero.

- `num % 2 == 0` → even
- Otherwise → odd

````python
num = int(input("Enter a number: "))

if num % 2 == 0:
    print(num, "is even")
else:
    print(num, "is odd")
````

## 6. Continuous `if-else` / `if-elif-else` Ladder

An if-else ladder checks multiple conditions in sequence. The first True condition's block is executed.

**Long form using nested `if-else`:**
```python
if condition1:
    statements
else:
    if condition2:
        statements
    else:
        if condition3:
            statements
        else:
            default_statements
```

**Simpler form using `elif`:**
```python
if condition1:
    statements
elif condition2:
    statements
elif condition3:
    statements
else:
    default_statements
```

`elif` means “else if”.

### Program: Positive, Negative or Zero

````python
num = float(input("Enter a number: "))

if num == 0:
    print(num, "is zero")
elif num > 0:
    print(num, "is positive")
else:
    print(num, "is negative")
````

### Program: Result Based on Percentage

| Percentage | Result |
|---|---|
| 85 to 100 | Distinction |
| 60 to below 85 | First class |
| 50 to below 60 | Second class |
| 35 to below 50 | Pass |
| Below 35 | Fail |

Values below 0 or above 100 are invalid.

````python
per = float(input("Enter percentage: "))

if per < 0 or per > 100:
    print("Invalid input")
elif per >= 85:
    print("Distinction")
elif per >= 60:
    print("First class")
elif per >= 50:
    print("Second class")
elif per >= 35:
    print("Pass")
else:
    print("Fail")
````

## 7. Triangle Classification

A triangle is possible only when:
- The sum of any two sides is greater than the third side.
- All side lengths are positive.

Types:
- **Equilateral:** all three sides are equal.
- **Isosceles:** exactly two sides are equal (or at least two, in the broad mathematical definition).
- **Scalene:** all three sides are different.

````python
s1 = float(input("Enter first side of triangle: "))
s2 = float(input("Enter second side of triangle: "))
s3 = float(input("Enter third side of triangle: "))

if s1 <= 0 or s2 <= 0 or s3 <= 0:
    print("Not a valid triangle")
elif s1 + s2 <= s3 or s1 + s3 <= s2 or s2 + s3 <= s1:
    print("Not a triangle")
elif s1 == s2 == s3:
    print("Equilateral triangle")
elif s1 == s2 or s1 == s3 or s2 == s3:
    print("Isosceles triangle")
else:
    print("Scalene triangle")
````

**Correction to the classroom version:** The original code used `<` instead of `<=` in the triangle inequality check, and used two separate `if` statements. That could print “Not a triangle” and then also classify it. The corrected version uses `elif` so only one result is printed.

## 8. Nested Conditional Statements

A nested conditional statement means placing one `if-else` statement inside another.

**Syntax:**
```python
if condition1:          # outer if
    if condition2:      # inner if
        statements
    else:
        statements
else:
    statements
```

### Program: Positive, Negative or Zero Using Nested `if`

````python
num = float(input("Enter a number: "))

if num != 0:
    if num > 0:
        print(num, "is positive")
    else:
        print(num, "is negative")
else:
    print(num, "is zero")
````

### Program: ATM Withdrawal Validation

Conditions:
1. Check whether the account balance is enough.
2. If enough, check whether the withdrawal amount is a multiple of 100.

This classroom example does not update the account balance; it only checks the conditions.

````python
account_balance = float(input("Enter account balance: "))
wd_amount = float(input("Enter withdrawal amount: "))

if wd_amount <= 0:
    print("Enter a valid withdrawal amount")
elif account_balance >= wd_amount:
    if wd_amount % 100 == 0:
        print("Transaction successful")
    else:
        print("Enter an amount in multiples of 100")
else:
    print("Insufficient balance")
````

### Program: Discount Calculation Using Nested `if`

Rules from the classroom example:
- Bill amount below Rs. 2,000: no discount.
- Rs. 2,000 to Rs. 5,000: 5% discount.
- Above Rs. 5,000 up to Rs. 10,000: 10% discount.
- Above Rs. 10,000: 20% discount.

````python
bill_amount = float(input("Enter bill amount: "))

if bill_amount >= 2000:
    if bill_amount > 5000:
        if bill_amount > 10000:
            discount = bill_amount * 0.20
        else:
            discount = bill_amount * 0.10
    else:
        discount = bill_amount * 0.05
else:
    discount = 0

amount_to_pay = bill_amount - discount

print("Discount amount:", discount)
print("Bill amount to pay:", amount_to_pay)
````

## 9. Day 2 – Quick Revision

- `if` executes a block when a condition is True.
- `else` executes when the `if` condition is False.
- `elif` checks another condition when previous conditions are False.
- A nested `if` is an `if` statement inside another conditional statement.
- `and` requires all conditions to be True.
- `or` requires at least one condition to be True.
- Use indentation consistently; Python uses indentation to define code blocks.
- Use `elif` when the program should choose only one result from several alternatives.

## 2. List in Python

A **list** is a collection that stores multiple values in a single variable. A list can contain values of different data types.

### Representation
- Square brackets: `[]`
- Or the `list()` constructor

### Properties
- **Heterogeneous:** can contain different data types.
- **Ordered:** items have a defined position.
- **Allows duplicates.**
- **Mutable:** items can be changed after creation.
- Supports indexing and slicing.


```python
l1 = [98, 123.34, "Python", True, 45 + 3j, True, 98]
print(l1)
print(type(l1))
```

```python
# Access an item using its index
l1[2]
```

### 2.1 Adding items to a list

- `append(value)` adds one item to the end.
- `insert(index, value)` adds an item at a specified position.


```python
l1 = [98, 123.34, "Python", True, 45 + 3j, True, 98]

l1.append("NANDANA P P")
print(l1)

l1.insert(3, 100)  # insert(index, value)
print(l1)
```

### 2.2 Updating an item

Because a list is mutable, an item can be changed using its index.

```python
l1 = [98, 123.34, "Python", True, 45 + 3j, True, 98]
l1[2] = "Java"
print(l1)
```

### 2.3 Removing items from a list

- `pop()` removes and returns the last item.
- `pop(index)` removes and returns the item at the given index.
- `remove(value)` removes the first matching value.
- `clear()` removes all items but keeps the list variable.
- `del list_name` deletes the variable itself.


```python
l1 = [98, 123.34, "Python", 100, True, 45 + 3j, True, 98]

l1.pop()
print(l1)

l1.pop(5)
print(l1)

l1.remove(True)
print(l1)

l1.clear()
print(l1)
```

```python
# del removes the list variable itself.
sample_list = [10, 20, 30]
del sample_list

# Trying to access sample_list after deletion raises NameError.
# print(sample_list)
```

### 2.4 Creating even and odd number lists

Create separate lists containing even and odd numbers from 1 to 20.

```python
even = []
odd = []

for i in range(1, 21):
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)

print("Even numbers:", even)
print("Odd numbers:", odd)
```

### 2.5 Create a square list

Create a new list containing the square of each value in the original list.

```python
l2 = [3, 2, 9, 5, 12, 10]
squares = []

for i in l2:
    squares.append(i * i)

print("Original list:", l2)
print("Square list:", squares)
```

### 2.6 Remove duplicates from a list

The following method keeps the first occurrence of each value and preserves the original order.

```python
l1 = [1, 4, 2, 5, 2, 7, 4, 9, 3]
unique_values = []

for i in l1:
    if i not in unique_values:
        unique_values.append(i)

print(unique_values)
```

### 2.7 Add corresponding elements of two lists

The lists below have the same length. Add the values at matching positions.

```python
l1 = [3, 7, 2, 8, 1]
l2 = [5, 4, 6, 9, 2]
output = []

for i in range(len(l1)):
    output.append(l1[i] + l2[i])

print(output)
```

**Note:** The last pair is `1 + 2`, so the last result is `3`. The correct output is `[8, 11, 8, 17, 3]`.


### 2.8 List operations on two lists

Given two lists, the examples below demonstrate:
- Concatenation: combine both lists, keeping all values.
- Unique combined values: combine without repeating values.
- Difference: values in the first list but not the second.
- Reverse difference: values in the second list but not the first.
- Common values: values present in both lists.
- Symmetric difference: values present in either list, but not in both.


```python
l1 = [1, 2, 3, 4, 5]
l2 = [4, 5, 6, 7, 8]

# 1. Concatenation
op1 = l1 + l2
print("Concatenation:", op1)

# 2. Combine unique values, preserving order
op2 = []
for i in l1:
    if i not in op2:
        op2.append(i)
for i in l2:
    if i not in op2:
        op2.append(i)
print("Unique combined values:", op2)

# 3. Values in l1 but not in l2
op3 = []
for i in l1:
    if i not in l2:
        op3.append(i)
print("Only in l1:", op3)

# 4. Values in l2 but not in l1
op4 = []
for i in l2:
    if i not in l1:
        op4.append(i)
print("Only in l2:", op4)

# 5. Common values
op5 = []
for i in l1:
    if i in l2:
        op5.append(i)
print("Common values:", op5)

# 6. Values that are in either list, but not both
op6 = []
for i in l1:
    if i not in l2:
        op6.append(i)
for i in l2:
    if i not in l1:
        op6.append(i)
print("Symmetric difference:", op6)
```

### 2.9 Combine strings at matching positions

The following example combines strings from two lists by their positions.

```python
list1 = ["M", "Na", "I", "NAND"]
list2 = ["y", "me", "s", "ANA P P"]
output = []

for i in range(len(list1)):
    output.append(list1[i] + list2[i])

print(output)
```

## 3. Tuple in Python

A **tuple** is a collection that stores multiple values in a single variable. It can contain different data types.

### Representation
- Parentheses: `()`
- Or the `tuple()` constructor

### Properties
- **Heterogeneous:** can contain different data types.
- **Ordered:** items have a defined position.
- **Allows duplicates.**
- **Immutable:** items cannot be changed after creation.
- Supports indexing and slicing.


```python
t1 = (98, 123.34, "Python", True, 45 + 3j, True, 98)

print(t1)
print(type(t1))
print(t1[2])
```

```python
# Create tuples using parentheses and tuple()
t1 = (10, 20, 30, 40, 50)
t2 = tuple([1, 2, 3, 4, 5])

print(t1)
print(t2)
```

### 3.1 Single-item tuple

A comma is required when creating a tuple with one item. Without the comma, parentheses simply group the value.

```python
single_item_tuple = (10,)
print(single_item_tuple)
print(type(single_item_tuple))

not_a_tuple = (10)
print(type(not_a_tuple))
```

### 3.2 Indexing

Tuple indexing works like list indexing:
- Positive indexes start at `0`.
- Negative indexes start at `-1` from the end.


```python
t1 = (10, 20, 30, 40, 50)

print(t1[0])
print(t1[2])
print(t1[-1])
print(t1[-2])
```

### 3.3 Slicing

Slicing extracts part of a tuple. The stop index is excluded.

```python
t1 = (10, 20, 30, 40, 50, 60)

print(t1[1:4])
print(t1[:3])
print(t1[2:])
print(t1[::2])
print(t1[::-1])
```

### 3.4 Tuple methods

- `count(value)` returns how many times a value occurs.
- `index(value)` returns the index of the first occurrence.


```python
t1 = (10, 20, 30, 20, 40, 20)

print("Count of 20:", t1.count(20))
print("First index of 20:", t1.index(20))
```

### 3.5 Built-in functions with tuples

For a tuple of numeric values, useful functions include:
- `len()` – number of items.
- `max()` – largest item.
- `min()` – smallest item.
- `sum()` – total of numeric items.
- `sorted()` – returns a new sorted **list**.


```python
t1 = (10, 20, 30, 40, 50)

print(len(t1))
print(max(t1))
print(min(t1))
print(sum(t1))
print(sorted(t1))
print(type(sorted(t1)))
```

### 3.6 Tuple packing and unpacking

**Packing:** assigning multiple comma-separated values creates a tuple.

**Unpacking:** assigning tuple items to separate variables.

```python
# Tuple packing
t1 = 10, 20, 30
print(t1)
print(type(t1))

# Tuple unpacking
a, b, c = t1
print(a)
print(b)
print(c)
```

### 3.7 Can we modify a tuple?

A tuple is immutable, so methods such as `append()`, `insert()`, `remove()` and `pop()` are not available. Item assignment is also not allowed.

If modification is needed, convert the tuple to a list, make the change, and convert it back to a tuple.

```python
t1 = (10, 20, 30)

# These operations are invalid for tuples:
# t1.append(40)
# t1[1] = 100
# t1.pop()

l1 = list(t1)
l1.append(40)
l1[0] = 5

t1 = tuple(l1)
print(t1)
```

### 3.8 Tuple concatenation

The `+` operator joins two tuples and creates a new tuple.

```python
t1 = (28, "Python", 45.93)
t2 = ("Java", 49)

combined = t1 + t2
print(combined)
```

### 3.9 Create even and odd tuples

Use temporary lists to collect values, then convert the lists into tuples.

```python
even = []
odd = []

for i in range(1, 21):
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)

even = tuple(even)
odd = tuple(odd)

print("Even numbers:", even)
print("Odd numbers:", odd)
```

### 3.10 Create a tuple of squares

```python
t1 = (3, 2, 9, 5, 12, 10)
squares = []

for i in t1:
    squares.append(i * i)

squares = tuple(squares)

print("Original tuple:", t1)
print("Square tuple:", squares)
```

### 3.11 Create pairs from two lists as a tuple

This example combines items at matching positions and stores each pair in a tuple.

```python
l1 = [3, 9, 2, 6, 4]
l2 = [1, 7, 9, 4, 8]

output = ()

for i in range(len(l1)):
    output = output + ((l1[i], l2[i]),)

print(output)
```

This example assumes both lists have the same length. If their lengths differ, use `zip()` or validate the lengths first.


## 4. Set in Python

A **set** is a collection used to store multiple values in a single variable.

### Representation
- Curly braces: `{}` for a non-empty set.
- Or the `set()` constructor.

**Important:** `{}` creates an empty dictionary, not an empty set. Use `set()` to create an empty set.

### Properties
- Can contain different hashable data types.
- Unordered: do not rely on a particular display order.
- Does not allow duplicate values.
- Mutable: items can be added or removed.
- Does not support indexing or slicing.
- Set elements must be hashable; lists and dictionaries cannot be elements of a set.


```python
s1 = {28, "Python", 45.93, True, "Program", 2 + 6j, "Python", False}

print(s1)
print(type(s1))

empty_set = set()
print(type(empty_set))
```

### 4.1 Adding items

Use `add(value)` to add one item to a set. Adding a value that is already present does not create a duplicate.

```python
s1 = {"Python", "Java", 100}
s1.add(200)
s1.add("Python")

print(s1)
```

### 4.2 Accessing set items

Sets do not support indexing because they are unordered. Therefore, `s1[4]` raises a `TypeError`.

To visit each item, use a loop.

```python
s1 = {"Python", "Java", "SQL"}

for item in s1:
    print(item)
```

### 4.3 Removing items

- `remove(value)` removes the specified value; it raises `KeyError` if the value is absent.
- `discard(value)` removes the value if present; it does not raise an error if absent.
- `pop()` removes and returns an arbitrary item. You should not assume which item will be removed.
- `clear()` removes all items.


```python
s1 = {"Python", "Java", "SQL", "Excel"}

s1.remove("Java")
print(s1)

s1.discard("Power BI")  # No error if the value is absent
print(s1)

removed_item = s1.pop()  # Arbitrary item; do not assume which one
print("Removed item:", removed_item)
print("Set after pop:", s1)
```

### 4.4 Set operations

Given two collections, set operations can find combined, different and common values.

- **Union:** all distinct elements from both sets.
- **Difference:** elements in the first set but not the second.
- **Intersection:** elements common to both sets.
- **Symmetric difference:** elements in either set, but not in both.


```python
l1 = [1, 2, 3, 4, 5]
l2 = [4, 5, 6, 7, 8]

s1 = set(l1)
s2 = set(l2)

print("Union:", s1.union(s2))
print("Difference (s1 - s2):", s1.difference(s2))
print("Difference (s2 - s1):", s2.difference(s1))
print("Intersection:", s1.intersection(s2))
print("Symmetric difference:", s1.symmetric_difference(s2))
```

### 4.5 Set operations using operators

The same common operations can be written using operators:

| Method | Operator | Meaning |
|---|---|---|
| `union()` | `\|` | All distinct elements |
| `intersection()` | `&` | Common elements |
| `difference()` | `-` | In first set, not second |
| `symmetric_difference()` | `^` | In either set, not both |


```python
s1 = {1, 2, 3, 4, 5}
s2 = {4, 5, 6, 7, 8}

print("Union:", s1 | s2)
print("Intersection:", s1 & s2)
print("Difference:", s1 - s2)
print("Symmetric difference:", s1 ^ s2)
```

### Important note about Boolean values and sets

In Python, `True` is equal to `1`, and `False` is equal to `0`. Because sets keep only unique values, a set cannot contain both `True` and `1` as separate elements (or both `False` and `0` as separate elements).


```python
example_set = {True, 1, False, 0, 2}
print(example_set)
print("Number of distinct elements:", len(example_set))
```

## 5. Comparison: List vs Tuple vs Set

| Property | List | Tuple | Set |
|---|---|---|---|
| Representation | `[]` | `()` | `{}` / `set()` |
| Ordered | Yes | Yes | No |
| Allows duplicates | Yes | Yes | No |
| Mutable | Yes | No | Yes |
| Indexing | Yes | Yes | No |
| Typical use | Changeable collection | Fixed collection | Unique values and set operations |

> For an empty collection, use `[]` for a list, `()` for a tuple, and `set()` for a set. `{}` is an empty dictionary.

## 8. Quick revision

- **List:** ordered, mutable, allows duplicates.
- **Tuple:** ordered, immutable, allows duplicates.
- **Set:** unordered, mutable, stores unique elements.
- `append()` adds to the end of a list.

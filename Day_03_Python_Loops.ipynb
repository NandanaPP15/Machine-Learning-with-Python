## 1. Introduction to Loops

A **loop** is used to execute a block of code repeatedly.

Python has two main types of loops:
- `for` loop
- `while` loop

### Repeating a statement without a loop

The same statement can be written many times, but this is repetitive and difficult to maintain.

````python
print("Good afternoon")
print("Good afternoon")
print("Good afternoon")
print("Good afternoon")
print("Good afternoon")
````

### Repeating a statement using string multiplication

The `*` operator can repeat a string. The `\n` character adds a new line.

````python
print("Good afternoon\\n" * 5)
````

## 2. `for` Loop with a Sequence

A `for` loop iterates through each item in a sequence, such as a string, list or tuple.

**Syntax:**
```python
for iterative_variable in sequence:
    statements
```

### Example 1: Display each character in a string

````python
name = "NANDANA ADARSH"

for ch in name:
    print(ch)
````

### Example 2: Display position and character

The variable `pos` starts at 0 and increases by 1 for each character.

````python
name = "NANDANA ADARSH"
pos = 0

for ch in name:
    print(f"Position: {pos} and character: {ch}")
    pos = pos + 1
````

### Example 3: Using `enumerate()`

`enumerate()` gives both the index and the item during iteration.

````python
name = "NANDANA ADARSH"

for item in enumerate(name):
    print(f"Position: {item[0]} and character: {item[1]}")
````

### Example 4: Loop through a list

````python
l1 = ["ADARSH", "NANDANA", "KEERTHANA", "SHYNA",
      "MANOHARAN", "SHYJA", "BALAN", "ASHNA"]

for name in l1:
    print(f"Hello good afternoon {name} - Happy Deepavali")
````

## 3. `for` Loop with `range()`

`range()` generates a sequence of numbers.

**Syntax:**
```python
range(start, stop, step)
```

- `start`: starting value; default is 0.
- `stop`: stopping boundary; the stop value is excluded.
- `step`: increment or decrement; default is 1.

Example: `range(1, 11, 1)` generates numbers 1 through 10.

````python
for i in range(1, 11, 1):
    print("i =", i)
    print("Hello")
````

### Example 1: Print numbers from 51 to 60

````python
for i in range(51, 61):
    print(i)
````

### Example 2: Print even and odd numbers from 101 to 110

````python
for i in range(101, 111):
    if i % 2 == 0:
        print(i, "= even number")
    else:
        print(i, "= odd number")
````

### Example 3: Print even numbers from 102 to 110

The step size is 2, so the loop moves from one even number to the next.

````python
for i in range(102, 111, 2):
    print(i, "= even number")
````

## 4. Programs Using `for` Loop

### Program 1: Perfect square numbers between 1 and 100

A number is a perfect square if its square root is a whole number.

````python
for num in range(1, 101):
    root = int(num ** 0.5)
    if root * root == num:
        print(num)
````

**Alternative method:** Generate squares directly.

````python
for i in range(1, 11):
    print(i * i)
````

### Program 2: Factorial of a given number

Factorial of a positive integer `n` is the product of all positive integers from 1 to `n`.

Example: `5! = 5 × 4 × 3 × 2 × 1 = 120`.

The factorial of 0 is 1.

````python
num = int(input("Enter a non-negative integer: "))

if num < 0:
    print("Factorial is not defined for negative numbers")
else:
    fact = 1
    for i in range(num, 0, -1):
        fact = fact * i

    print("The factorial of", num, "is", fact)
````

### Program 3: Multiplication table of a given number

Display the multiplication table from 1 to 10.

````python
num = int(input("Enter a number: "))

for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")
````

### Program 4: Check whether a number is prime

A prime number is greater than 1 and has exactly two factors: 1 and itself.

````python
num = int(input("Enter a number: "))

if num <= 1:
    print(num, "is not a prime number")
else:
    for i in range(2, num):
        if num % i == 0:
            print(num, "is not a prime number")
            break
    else:
        print(num, "is a prime number")
````

## 5. Loop Control Statements

Python has three commonly taught loop control statements:

- `break`: terminates the loop immediately.
- `continue`: skips the remaining statements in the current iteration and moves to the next iteration.
- `pass`: does nothing; it is a placeholder statement.

### 5.1 `break`

Stop the loop when `i` becomes 8.

````python
for i in range(1, 1000):
    if i == 8:
        break
    print(i)
````

### 5.2 `continue`

Skip the value 8 and continue the loop.

````python
for i in range(1, 10):
    if i == 8:
        continue
    print(i)
````

Skip the values 8, 9 and 10.

````python
for i in range(1, 15):
    if i == 8 or i == 9 or i == 10:
        continue
    print(i)
````

### 5.3 `pass`

`pass` is a placeholder. It does not skip an iteration or stop a loop; execution simply continues to the next statement.

````python
for i in range(1, 10):
    if i == 8:
        pass
    print(i)
````

### Prime number using `break` and loop `else`

The `else` block of a `for` loop runs only if the loop finishes without executing `break`.

````python
num = int(input("Enter a number: "))

if num <= 1:
    print(num, "is not a prime number")
else:
    for i in range(2, num):
        if num % i == 0:
            print(num, "is not a prime number")
            break
    else:
        print(num, "is a prime number")
````

## 6. `while` Loop

A `while` loop repeats a block of code as long as its condition is True.

**Syntax:**
```python
initialization
while condition:
    statements
    increment_or_decrement
```

Remember to update the loop variable. Otherwise, the loop may run forever.

````python
i = 1

while i <= 10:
    print(i)
    i = i + 1
````

### Example 1: Odd numbers between 1 and 20

````python
i = 1

while i <= 20:
    print(i)
    i = i + 2
````

### Example 2: Odd numbers on one line using `end`

````python
oddnum = 1

while oddnum <= 20:
    print(oddnum, end=" ")
    oddnum += 2
````

## 7. Programs Using `while` Loop

### Program 1: Swap two numbers using a temporary variable

````python
a = 10
b = 20

print(f"Before swap: a = {a}, b = {b}")

temp = a
a = b
b = temp

print(f"After swap: a = {a}, b = {b}")
````

### Program 2: Swap two numbers without a temporary variable

Python supports simultaneous assignment.

````python
a = 10
b = 20

print(f"Before swap: a = {a}, b = {b}")

a, b = b, a

print(f"After swap: a = {a}, b = {b}")
````

### Program 3: First 8 numbers of the Fibonacci series

Each number after the first two is the sum of the previous two numbers.

````python
num1 = 0
num2 = 1
i = 1

while i <= 8:
    print(num1)
    total = num1 + num2
    num1, num2 = num2, total
    i += 1
````

### Program 4: Sum of digits of a number

For 538: `5 + 3 + 8 = 16`.

- `% 10` gets the last digit.
- `// 10` removes the last digit.

````python
num = int(input("Enter a number: "))
temp = abs(num)
digit_sum = 0

while temp > 0:
    rem = temp % 10
    digit_sum += rem
    temp //= 10

print("The sum of the digits is:", digit_sum)
````

### Program 5: Check whether a number is a palindrome

A palindrome number reads the same forward and backward. Example: 121.

This version converts the input to an integer and reverses its digits.

````python
num = int(input("Enter a number: "))
temp = num
working_num = abs(num)
rev = 0

while working_num > 0:
    rem = working_num % 10
    rev = (rev * 10) + rem
    working_num //= 10

print(f"Number: {temp}")
print(f"Reversed number: {rev}")

if temp >= 0 and temp == rev:
    print(temp, "is a palindrome")
else:
    print(temp, "is not a palindrome")
````

## 8. Nested Loops

A **nested loop** is a loop inside another loop.

- Outer loop: commonly controls rows.
- Inner loop: commonly controls columns or repeated items within each row.

**Syntax:**
```python
for row in range(start, stop):
    for col in range(start, stop):
        statements
    statements_after_inner_loop
```

### Pattern 1: 5 × 5 star square

````python
for row in range(1, 6):
    for col in range(1, 6):
        print("*", end=" ")
    print()
````

### Pattern 2: Print 1 to 5 in each of 10 rows

````python
for row in range(1, 11):
    for col in range(1, 6):
        print(col, end=" ")
    print()
````

### Pattern 3: Continuous numbers from 1 to 15 in 3 rows and 5 columns

````python
n = 1

for row in range(1, 4):
    for col in range(1, 6):
        print(n, end="\t")
        n += 1
    print()
````

### Pattern 4: Repeated row number triangle

````python
for i in range(1, 6):
    for j in range(1, i + 1):
        print(i, end="\t")
    print()
````

### Pattern 5: Continuous number triangle

````python
n = 1

for i in range(1, 6):
    for j in range(1, i + 1):
        print(n, end="\t")
        n += 1
    print()
````

### Pattern 6: Increasing numbers in each row

````python
for i in range(1, 6):
    for j in range(1, i + 1):
        print(j, end="\t")
    print()
````

### Pattern 7: Multiplication grid with even products only

Print an even product; print a blank space when the product is odd.

````python
for i in range(1, 6):
    for j in range(1, 6):
        if (i * j) % 2 == 0:
            print(i * j, end="\t")
        else:
            print("", end="\t")
    print()
````

## 9. Nested `while` Loops

**Syntax:**
```python
initialize_outer
while outer_condition:
    initialize_inner
    while inner_condition:
        inner_statements
        update_inner
    outer_statements
    update_outer
```

The inner loop completes its iterations for each iteration of the outer loop.

### Pattern 1: 5 × 5 star square using nested `while`

````python
i = 1

while i <= 5:
    j = 1
    while j <= 5:
        print("*", end=" ")
        j += 1
    print()
    i += 1
````

### Pattern 2: Hollow square using nested `while`

Print stars on the boundary and spaces inside.

````python
i = 1

while i <= 5:
    j = 1
    while j <= 5:
        if i == 1 or j == 1 or i == 5 or j == 5:
            print("*", end=" ")
        else:
            print(" ", end=" ")
        j += 1
    print()
    i += 1
````

### Pattern 3: Multiplication tables from 6 to 10 using nested `while`

````python
i = 1

while i <= 10:
    j = 6
    while j <= 10:
        print(f"{j} x {i} = {j * i}", end="\t")
        j += 1
    print()
    i += 1
````

### Program: Prime numbers from 2 to 20 using nested `while`

````python
i = 2

while i <= 20:
    j = 2
    while j < i:
        if i % j == 0:
            break
        j += 1
    else:
        print(i, end="\t")
    i += 1
````

## 10. Pattern Programs Using Nested `for` Loops

### Pattern 1: Right-aligned star triangle

````python
for i in range(1, 6):
    for j in range(5, i, -1):
        print(" ", end="")
    for k in range(1, i + 1):
        print("*", end="")
    print()
````

### Pattern 2: Diamond pattern

````python
for i in range(1, 6):
    for j in range(5, i, -1):
        print(" ", end="")
    for k in range(1, 2 * i):
        print("*", end="")
    print()

for i in range(4, 0, -1):
    for j in range(5, i, -1):
        print(" ", end="")
    for k in range(1, 2 * i):
        print("*", end="")
    print()
````

### Pattern 3: Hollow diamond pattern

````python
for i in range(1, 6):
    j = 5
    while j > i:
        print(" ", end="")
        j -= 1

    k = 1
    while k <= 2 * i - 1:
        if k == 1 or k == 2 * i - 1:
            print("*", end="")
        else:
            print(" ", end="")
        k += 1
    print()

for i in range(4, 0, -1):
    j = 5
    while j > i:
        print(" ", end="")
        j -= 1

    k = 1
    while k <= 2 * i - 1:
        if k == 1 or k == 2 * i - 1:
            print("*", end="")
        else:
            print(" ", end="")
        k += 1
    print()
````

### Pattern 4: Alphabet triangle growing and shrinking

`ord()` returns the numeric code of a character. `chr()` converts a numeric code into a character. The code starts at 65, which represents `A`.

````python
ch = 65
i = 1

while i <= 5:
    for j in range(1, i + 1):
        print(chr(ch), end="")
        ch += 1
    print()
    i += 1

i = 4

while i >= 1:
    for j in range(1, i + 1):
        print(chr(ch), end="")
        ch += 1
    print()
    i -= 1
````

## 11. Combining `for` and `while` Loops

### Example 1: `while` loop inside a `for` loop

````python
for i in range(1, 4):
    j = 1
    while j <= 3:
        print(j, end=" ")
        j += 1
    print()
````

### Example 2: `for` loop inside a `while` loop

````python
i = 1

while i <= 3:
    for j in range(1, 4):
        print(j, end=" ")
    print()
    i += 1
````

## 1. What is a function?

A **function** is a reusable block of code that performs a specific task. Functions help divide a large program into smaller, manageable parts.

A function is defined using a function name followed by parentheses, such as `function_name()`.

### Why do we use functions?
- Increase code reusability.
- Improve readability and modularity.
- Make debugging and testing easier.

## 2. Types of functions

1. **Built-in functions** – functions provided by Python, such as `print()`, `input()`, `len()`, and `sum()`.
2. **User-defined functions** – functions created by a programmer using the `def` keyword.
3. **Lambda functions** – small, anonymous functions written using the `lambda` keyword.

### General syntax of a user-defined function

```python
def function_name(parameter1, parameter2):
    statements
    return value
```

- `def` starts the function definition.
- `function_name` is the name given to the function.
- Parameters receive values when the function is called.
- `return` sends a result back to the place where the function was called. It is optional.


## 3. Four common function patterns

Functions are often introduced using four combinations:

| Pattern | Takes input? | Returns a value? |
|---|---:|---:|
| Without input and without return | No | No |
| Without input and with return | No | Yes |
| With input and without return | Yes | No |
| With input and with return | Yes | Yes |

These are useful learning categories. Python functions can also have optional parameters, keyword arguments, and other combinations.


### 3.1 Function without input and without return

**Syntax**
```python
def function_name():
    statements

function_name()  # function call
```

The function does not receive arguments and does not explicitly return a result. It can still perform an action, such as displaying output.


````python
def add1():
    a = int(input("Enter a number: "))
    b = int(input("Enter another number: "))
    total = a + b
    print(f"The sum of {a} and {b} is {total}")

add1()
````

Call the same function again whenever you need it. Each call runs the function body again and asks for new input.

````python
add1()
````

### Local variables

Variables created inside a function are usually **local variables**. They can be used inside that function, but they cannot normally be accessed directly outside it.

For example, after calling `add1()`, trying to run `print(a)` outside the function raises a `NameError`, because `a` is local to `add1()`.


````python
# This will raise NameError because a is local to add1().
# Uncomment the next line to see the error:
# print(a)
````

### Global variables

A **global variable** is created outside a function. It can be read inside functions in the same module.

Use `global` inside a function only when you need to reassign a global variable. In general, passing values as parameters and returning results makes functions easier to understand and test.


````python
message = "Hello from the global scope"

def show_message():
    print(message)

show_message()
print(message)
````

### 3.2 Function without input and with return

This function does not receive arguments, but it returns one or more values.

**Syntax**
```python
def function_name():
    statements
    return value

result = function_name()
```

A function can return multiple values. Python packages them into a tuple, which can be unpacked into separate variables.


````python
def add2():
    a = int(input("Enter a number: "))
    b = int(input("Enter another number: "))
    total = a + b
    return a, b, total

add2()
````

The returned tuple can be stored in one variable, or unpacked into three variables.

````python
x, y, result = add2()
print(f"The sum of {x} and {y} is {result}")
````

### 3.3 Function with input and without return

This function receives values through parameters and performs an action, but it does not explicitly return a result.

**Syntax**
```python
def function_name(parameter1, parameter2):
    statements

function_name(value1, value2)
```


````python
def add3(n1, n2):
    total = n1 + n2
    print(f"The sum of {n1} and {n2} is {total}")

add3(39,  38)
````

You can also collect values from the user first and then pass them as arguments.

````python
a = int(input("Enter a number: "))
b = int(input("Enter another number: "))

add3(a, b)
````

**Important:** `add3()` has two parameters, so it expects two arguments. Calling `add3(20, 30, 40)` gives a `TypeError` because three arguments were supplied.

A function call must provide the required number of arguments unless the function definition allows optional or variable numbers of arguments.


````python
# This call produces a TypeError because add3 expects 2 arguments.
# Uncomment to test:
# add3(20, 30, 40)
````

### 3.4 Function with input and with return

This function receives values through parameters, processes them, and returns a result.

**Syntax**
```python
def function_name(parameter1, parameter2):
    statements
    return result

result = function_name(value1, value2)
```


````python
def add4(a, b):
    total = a + b
    return total

x = int(input("Enter a number: "))
y = int(input("Enter another number: "))

result = add4(x, y)
print(f"The sum of {x} and {y} is {result}")
````

### Difference between `print()` and `return`

- `print()` displays information on the screen.
- `return` sends a value back to the caller so it can be stored, reused, or passed to another function.

A function without an explicit `return` statement returns `None` by default.


````python
def show_sum(a, b):
    print(a + b)

def get_sum(a, b):
    return a + b

shown = show_sum(5, 3)
returned = get_sum(5, 3)

print("Value returned by show_sum:", shown)
print("Value returned by get_sum:", returned)
````

## 4. Lambda functions

A **lambda function** is a small, anonymous (nameless) function in Python.

### Features
- It is defined using the `lambda` keyword instead of `def`.
- It can accept any number of arguments.
- It contains one expression.
- The expression's result is returned automatically; no `return` keyword is needed.

### Syntax
```python
lambda arguments: expression
```

### Why use lambda?
- Useful for short, simple operations.
- Can make a small function concise.
- Often used with functions such as `map()`, `filter()`, and `reduce()`, or as an inline function.

### When not to use lambda
- Avoid it for complex logic.
- Prefer `def` when a named function or clearer multi-line logic improves readability.


### Example 1: Addition using lambda

````python
add = lambda a, b: a + b

add(10, 30)
````

### Example 2: Check even or odd using lambda

The conditional expression has the form:

`value_if_true if condition else value_if_false`


````python
even_odd = lambda n: "even" if n % 2 == 0 else "odd"

even_odd(889)
````

### Example 3: More lambda practice

````python
square = lambda n: n ** 2
square(6)
````

````python
maximum = lambda a, b: a if a > b else b
maximum(18, 25)
````

## 13. Python Operators

### 13.1 Arithmetic operators

`+` addition, `-` subtraction, `*` multiplication, `/` division, `//` floor division, `%` modulus and `**` exponentiation.

````python
a = 7
b = 2

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
````

### 13.2 Relational (comparison) operators

These compare values and return `True` or `False`: `==`, `!=`, `>`, `<`, `>=`, `<=`.

````python
print(5 > 7)
print(5 == 5)
print(5 != 7)
````

### 13.3 Assignment operators

Assignment operators update a variable's value.

````python
n = 10
print(n)

n += 7   # same as n = n + 7
print(n)

n -= 5   # same as n = n - 5
print(n)

n **= 2  # same as n = n ** 2
print(n)
````

### 13.4 Logical operators

- `and` – True when both conditions are True.
- `or` – True when at least one condition is True.
- `not` – reverses the Boolean result.

````python
print(4 == 6 and 5 < 10)
print(4 == 6 or 5 < 10)
print(not True)
print(not False)
````

### 13.5 Membership operators

`in` checks whether a value is present in a collection. `not in` checks whether it is absent.

````python
l1 = [3, 7, 9, 1, 5]

print(9 in l1)
print(4 in l1)
print(9 not in l1)
````

### 13.6 Identity operators

`is` and `is not` check whether two references point to the same object. They are not substitutes for `==` and `!=`, which compare values.

Avoid examples such as `5 is 10`; Python may show a warning. Use identity checks mainly when comparing with `None` or when object identity is specifically important.

````python
x = [1, 2]
y = x
z = [1, 2]

print(x == z)  # same values
print(x is y)  # same object
print(x is z)  # different objects
````

### 13.7 Bitwise operators

Bitwise operators work on the binary representation of integers.

For example:

```text
12 = 1100
10 = 1010
     ----
12 & 10 = 1000 = 8
12 | 10 = 1110 = 14
```

- `&` bitwise AND
- `|` bitwise OR

````python
print(12 & 10)
print(12 | 10)
````

## 15. Simple Interest

Formula:

**Simple Interest = (Principal × Rate × Time) / 100**

**Total Amount = Principal + Simple Interest**

````python
principal = float(input("Enter principal amount: "))
rate = float(input("Enter rate of interest: "))
time = float(input("Enter number of years: "))

simple_interest = (principal * rate * time) / 100
total_amount = principal + simple_interest

print(f"Simple interest = Rs. {simple_interest:.2f}")
print(f"Total amount = Rs. {total_amount:.2f}")
````

## 16. Employee Salary Slip Program

The following program follows the formulas written in the classroom assignment:

- DA = 8% of Basic Salary
- TA = 12% of Basic Salary
- Bonus = Rs. 3,000
- PF = 15% of Basic Salary
- Tax = 5% of Basic Salary

**Formula clarification:** The assignment labels `Basic Salary + DA + TA + Bonus` as Net Salary and then adds PF and Tax to get Gross Salary. In common payroll terminology, gross earnings are usually calculated before deductions, while net salary is calculated after deductions. The code below preserves the trainer's stated formulas but labels them as specified.

````python
empid = input("Enter Employee ID: ")
ename = input("Enter Employee Name: ")
basic_salary = float(input("Enter Basic Salary: "))

da = basic_salary * 0.08
ta = basic_salary * 0.12
bonus = 3000
pf = basic_salary * 0.15
tax = basic_salary * 0.05

net_salary_as_given = basic_salary + da + ta + bonus
gross_salary_as_given = net_salary_as_given + pf + tax

print("*" * 10, "SALARY SLIP", "*" * 10)
print(f"EMP ID: {empid}")
print(f"EMP NAME: {ename}")
print(f"BASIC SALARY: {basic_salary:.2f}")
print(f"DAILY ALLOWANCE (8%): {da:.2f}")
print(f"TRAVEL ALLOWANCE (12%): {ta:.2f}")
print(f"BONUS AMOUNT: {bonus:.2f}")
print(f"PF (15%): {pf:.2f}")
print(f"TAX (5%): {tax:.2f}")
print(f"NET SALARY (as per assignment): {net_salary_as_given:.2f}")
print(f"GROSS SALARY (as per assignment): {gross_salary_as_given:.2f}")
print("*" * 10, "THANK YOU", "*" * 10)
````
